##### Descrição do Problema

Somos uma empresa de consultoria especializada em Engenharia e Análise de Dados e fomos contratados pela Agência Nacional de Aviação Civil (ANAC) para auxiliar em uma iniciativa de analisar dados voltamos ao programa de qualidade da agência, com foco em traçar planos de ação para **reduzir riscos da operação relacionadas a aviões**.

Entretanto, a empresa tem uma certa dificuldade em relação a organização e preparação dos dados para análise, hoje eles conseguem extrair esses dados manualmente de um sistema, que muitas vezes é preenchido de forma manual pelos funcionários, ocasionando em bases que não tem uma qualidade e consistencia suficiente para suportar as análises.

Com intuito de direcionar esse projeto, foi disponibilizada a base bruta e foram passadas algumas perguntas que o orgão gostaria de ter mais clareza sobre, para assim, através dos dados, conseguirem traçar melhor quais ações devem ser feitas, sendo elas:

**Pergunta 1:** `Quais causas de acidentes geram maior gravidade (com maior número de feridos)? Para assim conseguirmos direcionar quais causas devemos investigar e tentar ao máximo mitigar.`

**Pergunta 2:** `Quais os fabricantes ou modelos de aeronaves com maior incidencia de acidentes relacionadas a questões técnicas/de infraestrutura do avião.`

**Pergunta 3:** `Quais aeroportos tem maiores incidentes relacionados a funcionários/erros humanos? Com intuito de priorizarmos treinamentos nesses locais.`

###### Seleção dos Dados

Os dados utilizados foram obtidos no Kaggle, referente ao Dataset de Acidentes no Espaço Aéreo Brasileiro (Cenipa), segue o link para o dataset: https://www.kaggle.com/datasets/markfinn1/acidentes-areos/data?select=Cenipa.csv
