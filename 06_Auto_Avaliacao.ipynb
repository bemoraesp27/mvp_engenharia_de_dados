###### Autoavaliação

✅ **Objetivos:**
- Conseguimos aringir o objetivo inicial de responder as 3 perguntas de negócio inicialmente levantadas e estruturar um modelo de arquitetura de medalhão no databricksm

⚠️**Principais dificuldade:**
- Tinhamos muitos registros nulo variados (***, ####, NULL textual)
- Cerca de 81% das ocorrências não têm fator causal preenchido, então a análise de causas cobre só os casos com investigação publicada.
- Acidentes graves foi medido por fatalidades, porque a base não tem coluna de feridos, o que poderia enriquecer a análise.
- Aeródromo tem muitos nulos (mais de 2 mil registros sem informação), o que limita a hipótese 3.

💡**Possíveis incrementos no pipeline:**
- Na camada bronze foi realizada a ingestão dos dados de forma manual, poderia haver alguma forma de conexão direta com os dados, principalmente para novos registros e assim continuar alimentando esse pipeline com dados atuais, contendo testes automatizados sobre a qualidade dos dados, já que tivemos algumas questões relacionadas a isso.
- Poderia ter alguma ferramneta de orquestração para executar as etapas de cada camada do medalhão.
- Além da camada analyze poderia ter algum dashboard ou forma de consulta mais visual dos dados tratados, conectado diretamente a uma tabela gold um pouco maior do que as 3 criadas especificamente para cada resposta (como uma tabela "gold" focada em análise exploratória).
