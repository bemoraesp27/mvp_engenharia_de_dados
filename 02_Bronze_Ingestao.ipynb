###### Ingestão de dados brutos na camada Bronze

Neste notebook iremos realizar a ingestão de dados em nossa camada bronze para conseguir visualizar os dados dentro de nosso ambiente databricks e, assim, trabalharmos neles.

⚠️ **Observação:** Para fins de MVP, o csv contendo os dados que serão trabalhados foram ingestados de forma manual no schema bronze criado, assim, sendo criado um volume nomeado de "raw_files".

A seguir, iremos fazer a leitura desse csv e transformá-lo no formato de tabela para assim conseguirmos trabalhar nele!

```python
import re

df = spark.read.csv(
    "/Volumes/mvp_bernardomoraes_catalog/bronze/raw_files/Cenipa.csv",
    sep=";",
    header=True,
    inferSchema=True,
    encoding="UTF-8"
)

# substitui espaços e caracteres inválidos por "_"
novas_colunas = [re.sub(r'[ ,;{}()\n\t=]+', '_', c).strip('_') for c in df.columns]
df = df.toDF(*novas_colunas)

df.write.mode("overwrite").saveAsTable("mvp_bernardomoraes_catalog.bronze.ocorrencias_cenipa")
```

🔍Após a leitura do csv e escrita em formato de tabela, iremos visualizar sua quantidade total de linhas e colunas existentes:

```python
df_check = spark.table("mvp_bernardomoraes_catalog.bronze.ocorrencias_cenipa")

print("Linhas:", df_check.count())
print("Colunas:", len(df_check.columns))
df_check.printSchema()
```

📌Por fim, como boa prática, iremos adicionar uma descrição a tabela criado:

```python
%sql
COMMENT ON TABLE mvp_bernardomoraes_catalog.bronze.ocorrencias_cenipa 
IS 'Dados brutos de ocorrências aeronáuticas do CENIPA, sem tratamento. Fonte: CSV baixado manualmente e ingestado via Volume.';
```

```python
%sql
--Visualizando a tabela criada
SELECT * FROM mvp_bernardomoraes_catalog.bronze.ocorrencias_cenipa LIMIT 10;
```
