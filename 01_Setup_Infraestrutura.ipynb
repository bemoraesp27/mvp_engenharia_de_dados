###### Preparação do Setup de Infraestrutura

Neste notebook iremos focar na criação de todo o setup que irá suportar o nosso pipeline, desde o momento de ingestão de dados até a camada de resposta das perguntas traçadas no início do trabalho, já com os dados tratados.

Antes de tudo, vale ressaltar os seguintes conceitos de databricks:
- `Catalog:` sendo o nível mais alto, como um grande "contêiner" que agrupa tudo relacionado ao nosso projeto, área de negócio ou ambiente;
- `Schema:` sendo o nível dentro do Catalog, onde é agrupado as tabelas/views/volumes relacionados entre si;
- `Volume:` nível mais baixo, onde se encontram os dados de fato.

Além disso, foi utilizado a arquitetura de medalhão neste projeto, contendo as seguintes camadas:
- `Bronze`: sendo a camada onde os dados brutos são ingestados;
- `Silver:`camada de tratamento, limpeza e padronização dos dados brutos;
- `Gold:`sendo a camada de prepapração para a análise, onde são feitas agregagações e preparações que suportem as perguntas que querem ser respondidas;
- `Analyze:` ambiente em que serão criadas as tabelas/views que de fato irão responder as perguntas inicias.

```python
%sql

--Nesta etapa de código será criado o catalogo e schemas necessários para o armazenamento dos dados
CREATE CATALOG IF NOT EXISTS mvp_bernardomoraes_catalog;

USE CATALOG mvp_bernardomoraes_catalog;

CREATE SCHEMA IF NOT EXISTS bronze;
CREATE SCHEMA IF NOT EXISTS silver;
CREATE SCHEMA IF NOT EXISTS gold;
CREATE SCHEMA IF NOT EXISTS analyze;
```

📌A partir disso, já com o Catalog e Schemas criados, como boa prática, iremos adicionar descrições para cada um desses artefatos.

```python
%sql

--Criação das descrições referentes ao Catalog e a cada Schema criado na etapa anterior
COMMENT ON CATALOG mvp_bernardomoraes_catalog IS 'Catalog do MVP de análise de ocorrências aeronáuticas do CENIPA';

COMMENT ON SCHEMA bronze IS 'Dados brutos da extração da base do Kaggle descritas no notebook "00_Descrição_Problema". Nesta etapa será realizada e ingestão dos dados brutos.';
COMMENT ON SCHEMA silver IS 'Dados limpos, tipados e padronizados. Nesta etapa será realizada a transformação dos dados brutos';
COMMENT ON SCHEMA gold IS 'Dados agregados e prontos para consumo final do time de Business Intelligence';
COMMENT ON SCHEMA analyze IS 'Tabelas e views para apoio na reposta das perguntas direcionadas no início do trabalho';
```

```python
%sql
--Visualizando os schemas criados para confirmar se o processo foi realizado corretamente
SHOW SCHEMAS IN mvp_bernardomoraes_catalog;
```

📌 Agora que já realizamos a criação de todo o ambiente que iremos precisar para este MVP, vamos criar um volume destinado a realizar o upload do arquivo csv bruto que extraímos do Kaggle.

Para isso, dentro da nossa camada bronze, criaremos um volume apelidado de "raw_files", uma boa prática de nomenclatura para definir arquivos brutos.

```python
%sql

--Criando artefato apara realizarmos o upload do arquivo bruto
CREATE VOLUME IF NOT EXISTS mvp_bernardomoraes_catalog.bronze.raw_files;
```

✅ Finalizada toda a criação do setup, podemos partir para a ingestão dos dados brutos na camada bronze!
