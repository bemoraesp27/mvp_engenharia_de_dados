###### Tabelas agrupadas na camada Gold

Neste notebook iremos realizar a criação de tabelas agrupadas que serão espefícias para responder cada uma de nossas hipóteses iniciais, assim, os times de negócio conseguem realizar consultas e visualizar as informações necessárias para responder suas perguntas de forma objetiva e clara.

Relembrando nossas hipóteses iniciais:

**Pergunta 1:** `Quais causas de acidentes geram maior gravidade (com maior número de feridos)? Para assim conseguirmos direcionar quais causas devemos investigar e tentar ao máximo mitigar.`

**Pergunta 2:** `Quais os fabricantes ou modelos de aeronaves com maior incidencia de acidentes relacionadas a questões técnicas/de infraestrutura do avião.`

**Pergunta 3:** `Quais aeroportos tem maiores incidentes relacionados a funcionários/erros humanos? Com intuito de priorizarmos treinamentos nesses locais.`

```python
%sql
CREATE OR REPLACE TABLE mvp_bernardomoraes_catalog.gold.causas_acidentes_fatalidades AS
SELECT
    fator_acidente,
    area_fator,
    COUNT(*) AS qtd_acidentes,
    SUM(total_fatalidades_acidente) AS total_fatalidades
FROM mvp_bernardomoraes_catalog.silver.ocorrencias_cenipa
WHERE classificacao_ocorrencia = 'ACIDENTE'
GROUP BY fator_acidente, area_fator
ORDER BY total_fatalidades DESC;

COMMENT ON TABLE mvp_bernardomoraes_catalog.gold.causas_acidentes_fatalidades 
IS 'Causas (fator_acidente/area_fator) e sua relação com fatalidades em acidentes, respondendo: quais causas mais geram acidentes graves?';
```

```python
%sql
CREATE OR REPLACE TABLE mvp_bernardomoraes_catalog.gold.fabricante_modelo_falhas_tecnicas AS
SELECT
    fabricante_aeronave,
    modelo_aeronave,
    tipo_ocorrencia,
    COUNT(*) AS qtd_ocorrencias
FROM mvp_bernardomoraes_catalog.silver.ocorrencias_cenipa
WHERE tipo_ocorrencia IN (
    'FALHA DO MOTOR EM VOO', 'FALHA DO MOTOR NO SOLO',
    'FALHA OU MAU FUNCIONAMENTO DE SISTEMA / COMPONENTE',
    'FALHA ESTRUTURAL', 'COM TREM DE POUSO', 'COM COMANDOS DE VOO',
    'COM PARA-BRISAS / JANELA / PORTA', 'DESCOMPRESSÃO NÃO INTENCIONAL / EXPLOSIVA',
    'FUMAÇA NA CABINE', 'ESTOURO DE PNEU'
)
GROUP BY fabricante_aeronave, modelo_aeronave, tipo_ocorrencia
ORDER BY qtd_ocorrencias DESC;

COMMENT ON TABLE mvp_bernardomoraes_catalog.gold.fabricante_modelo_falhas_tecnicas 
IS 'Fabricante e modelo de aeronave por ocorrência de natureza técnica (falha de motor, sistema, estrutural etc.)';
```

```python
%sql
CREATE OR REPLACE TABLE mvp_bernardomoraes_catalog.gold.aerodromo_fator_humano AS
SELECT
    aerodromo_ocorrencia,
    uf_ocorrencia,
    area_fator,
    COUNT(*) AS qtd_ocorrencias
FROM mvp_bernardomoraes_catalog.silver.ocorrencias_cenipa
WHERE area_fator IN ('FATOR HUMANO', 'FATOR OPERACIONAL')
  AND aerodromo_ocorrencia IS NOT NULL
GROUP BY aerodromo_ocorrencia, uf_ocorrencia, area_fator
ORDER BY qtd_ocorrencias DESC;

COMMENT ON TABLE mvp_bernardomoraes_catalog.gold.aerodromo_fator_humano 
IS 'Aeródromo com maior incidência de ocorrências ligadas a fator humano/operacional';
```
